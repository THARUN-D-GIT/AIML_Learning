```python
#takes 3 input and prints their averages
def Average(a,b,c):
    return (a+b+c)/3
print(Average(1.0,2.7,3.8))
```

```python
#default value for parameters
def para(a,b=0):
    return a+b;
print(para(7,6),para(9))
```

```python
#Lambda expressions
'''
Lambda expression can have mulitple parameters but can have only one results 
Expression:which is described as result above which will be like mathematcal equation
'''
avg=lambda a,b,c: (a+b+c)/3
print(avg(2,6,4))
```

```python
#WAP to print any factorial of a number
n=int(input("enter the value"))
def cal_fact(n):
    fact=1
    for i in range(1,n+1):
        fact*=i
    print(fact)
cal_fact(n)
```
