```python
#Loops 
#while loops
i=1;
j=1;
while(i<=5):
    print("Hello world")
    i+=1
while(j<=5):
    print(j)
    j+=1
```

```python
#print multiplication of any number
num=int(input("enter the value of num: "))
i=1
while(i<=10):
    print(num*i)
    i+=1

```

```python
# break and continue
#break to stop or terminate the loop
# contiue is to ignore any itteration
i=0
while(i<=10):
    i+=1
    if(i%3==0):
        continue
    if(i==7):
        break
    print(i)
    
    
```

```python
 #for loop
string="Hello"
for var in string:
    print(var)
if 'o' in string:
    print("exists")
```

```python
# range always print number from 0 to n-1 where n is the upper bound
for i in range(10):
    print(i)
```

```python
word="Artificial intelligence"
ans=0
for ch in word:
    if(ch=='i'):
        ans+=1;
print(ans)
```

```python
#print vowel count in a string
cnt=0
word=input("enter any word")
for ch in word:
    if(ch=='a' or ch=='e' or ch=='i' or ch=='o' or ch=='u'):
        cnt+=1
print("number of vowel in",word,"is:",cnt)
```

```python
#Range Function
#range (5) 0-4
'''
range(start,stop,step)
stop value is always compulsary
if no value is entered for start default value is 0
Step will increase by 1
'''

for i in range(10):
    print(i)
for j in range(1,10,2):
    print(j)
```

```python
#print sum of first n natural numbers
n=int(input("enter value of n"))
sum=0
for i in range(1,n+1):
    sum+=i
print(sum)
```
