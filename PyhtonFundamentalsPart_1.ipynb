```python
#Match case
#Similar to switch case in java but here we use match instead of switch 
#and also for default we will use case _
color=input("enter the color ur seeing")
match color:
    case "Green":
        print("get lost")
    case "Yellow":
        print("wait")
    case "Red":
        print("STFU")
    case _:
        print("Are u dumb bro")
        
```
