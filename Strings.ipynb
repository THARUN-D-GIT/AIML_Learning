```python
#Strings
'''
len(Strings)
gives the length of the strings
'+': concatenets the 2 string 
indexing is used to acceses each character like any other language
Immutable:Strings cant be altered 
'''
word1="Hello"
word2="Python"
print(len(word1),len(word2))
print(word1+" "+word2)
for i in range(0,len(word1)):
    print(word1[i])
#slicing 
'''
In Slicing we use String[Statrt:End] index if end index is not specified it will 
print untill the end of String

we can use negative indexing to print the strings in reverse order
p  y  t  h  o  n
0  1  2  3  4  5
-6 -5 -4 -3 -2 -1
'''
word="I am learning ai and ml"
print(len(word))
print(word[12:22]) 
print(word2[-4:-2])
```
