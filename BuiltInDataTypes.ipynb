```python
#Built in Data types in Python
#List 
'''
List: it is a mutable sequence of values
'''
marks=[99,89,65,98,25]
print(marks)
print(len(marks))
print(marks[1])
print("Before:",marks[4])
marks[4]=69
print("After:",marks[4])

# gives error print(marks[87])
#sLicing 
'''
list[start:end] end is not included
'''
print(marks[0:3])
```

```python
#Mehtods used in Lists
'''
1.
l.append():adds value to the end
2.
l.insert(idx,val):adding value to a particular index 
it doesnot removes any existing vlaues it just moves every value after given index by one
3.
l.sort(): sort the lists
4.
l.reverse()
reverses the list
'''
l=[1,2,3,4,5,6]
print(l)
l.append(7)
print(l)
l.insert(2,8)
print(l)
l.sort()
print(l)
l.reverse()
print(l)
```

```python
#tuples
'''
Tupples:Immutable sequences of value
similar to lists but cant edit the value
we can use any data types in lists
we use "()" to assign values in a tupple
''' 
tup=(1,2,4,5,6,8)
# gives error cant edit tupple valuetup[3]=8
print(tup)
print(type(tup))
print(tup[3])
'''
if we want to create single value tupple then we cant use 
tup=(13)
it will give data types as int 
instead we will use tup(13,)
then it will be single value tupple 
first will be treated as varible
'''
tup1=(3)
tup2=(3,)
print(type(tup1),type(tup2))
```

```python
#Built in methods for tupples
'''
t.index(val):
returns the index of first occurance of val
t.count(val):
returns the total occurance of val

'''
t=(1,2,4,2,5,2,6)
print(t.index(2))
print(t.count(2))
```

```python
#DICTIONARY
'''
it contains key:value pair
keys are unique
use {} curley braces to create the dictionary
use quotes as per their respective data type
we can acceses the vlaue of dict using keys
dictionarty`s are mutable
'''
info={
    "name":"Tarun",
    "cgpa":8.79,
    "Languages":"Pyhton,Java",
    "sem":5    
}
print(type(info))
print(info["name"],info["cgpa"])
info["cgpa"]=8.98
print(info["name"],info["cgpa"])
```

```python
#Methods
'''
d.keys(): Returns all the keys
d.values():Returns all the vlaue
d.items(): returns key value pair
d.get(val):returns value according to keys
d.update(new_item):adds new item to dict
'''
print(info.keys())
dict_keys=list(info.keys())
print(type(dict_keys))
print(info.values()) 
print(info.items())
#print(info["cgpa2"]) #gives error:Key error
print(info.get("cgpa")) #dont gives error returns none if key is not present
info.update({
    "city":"Delhi"
})
print(info.items())
```
